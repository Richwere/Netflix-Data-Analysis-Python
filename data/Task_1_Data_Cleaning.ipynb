```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os
```

```python

# Load the dataset using its absolute file path
file_path = "/home/richard/Netflix_Fresh/data/raw/Dataset.csv"

df = pd.read_csv(file_path)

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

df.head()
```

```python

# Display column names and data types
print("Dataset information:")
df.info()

# Check the dataset dimensions
print("\nDataset shape:", df.shape)

# Display statistical summary
df.describe(include="all").T
```

```python

# Count missing values in each column
print("Missing values per column:")
print(df.isnull().sum())

# Count fully duplicated rows
print("\nNumber of duplicate rows:", df.duplicated().sum())

# Check repeated titles
print("\nNumber of repeated title entries:", df["title"].duplicated().sum())

# Inspect placeholder values
print("\n'Not Given' counts:")
for column in ["director", "country", "rating"]:
    print(f"{column}: {(df[column] == 'Not Given').sum()}")
```

```python

# Display titles that appear more than once
repeated_titles = df[df["title"].duplicated(keep=False)].sort_values("title")

repeated_titles[["show_id", "title", "type", "release_year", "country"]]
```

```python

# Check the original column names
print("Column names:")
print(df.columns.tolist())

# Display selected records without changing the dataset
display(df.iloc[[2925, 4261, 3285, 4260, 393, 537]])

# Check the first few records
display(df.head())
```

```python

# Compare duplicate titles, excluding the unique show_id
comparison_columns = [
    "type", "title", "director", "country", "date_added",
    "release_year", "rating", "duration", "listed_in"
]

repeated_titles = df[df["title"].duplicated(keep=False)]

for title, group in repeated_titles.groupby("title"):
    print(f"\nTitle: {title}")
    display(group[comparison_columns])
    print(
        "All other columns identical:",
        group[comparison_columns].drop_duplicates().shape[0] == 1
    )
```

```python

# Record the number of rows before cleaning
rows_before = len(df)

# Remove exact duplicates across all columns except the unique show_id
df = df.drop_duplicates(subset=[
    "type", "title", "director", "country", "date_added",
    "release_year", "rating", "duration", "listed_in"
])

# Check the result
rows_after = len(df)

print("Rows before cleaning:", rows_before)
print("Rows after cleaning:", rows_after)
print("Duplicate records removed:", rows_before - rows_after)
```

```python

# Standardize whitespace in text columns
text_columns = df.select_dtypes(include="object").columns

for column in text_columns:
    df[column] = df[column].str.strip()

# Replace placeholder values with actual missing values
df[["director", "country"]] = df[["director", "country"]].replace(
    "Not Given", np.nan
)

# Check the results
print("Missing values after standardization:")
print(df.isnull().sum())

print("\nSample standardized records:")
display(df.head())
```

```python

# Standardize content type and rating labels
df["type"] = df["type"].str.strip().str.title()
df["rating"] = df["rating"].str.strip().str.upper()

# Check unique values
print("Content types:")
print(df["type"].value_counts())

print("\nRating categories:")
print(sorted(df["rating"].dropna().unique()))

# Check for missing ratings
print("\nMissing ratings:", df["rating"].isna().sum())
```

```python

# Restore the standard Netflix content-type labels
df["type"] = df["type"].replace({
    "Tv Show": "TV Show",
    "Movie": "Movie"
})

# Verify the final labels and counts
print(df["type"].value_counts())
```

```python

# Convert date_added from text to datetime
df["date_added"] = pd.to_datetime(
    df["date_added"],
    format="%m/%d/%Y",
    errors="coerce"
)

# Check the results
print("Data type:", df["date_added"].dtype)
print("Missing dates:", df["date_added"].isna().sum())

print("\nEarliest date added:", df["date_added"].min())
print("Latest date added:", df["date_added"].max())

display(df[["title", "date_added"]].head())
```

```python

# Identify titles that look like abbreviated dates
date_like_titles = df[
    df["title"].str.fullmatch(
        r"\d{1,2}-[A-Za-z]{3}",
        na=False
    )
]

print("Number of date-like titles:", len(date_like_titles))
display(
    date_like_titles[
        ["show_id", "title", "type", "release_year", "country"]
    ]
)
```

```python

# Final data-quality checks
print("Final dataset shape:", df.shape)

print("\nDuplicate rows:", df.duplicated().sum())

print("\nMissing values per column:")
print(df.isna().sum())

print("\nUnique content types:")
print(df["type"].unique())

print("\nUnique show IDs:", df["show_id"].nunique())

print("\nDate-like titles retained:", len(date_like_titles))
```

```python

# Save the cleaned dataset
output_path = "/home/richard/Netflix_Fresh/data/cleaned/netflix_cleaned.csv"

df.to_csv(output_path, index=False)

print("Cleaned dataset saved successfully!")
print("File location:", output_path)
print("Rows saved:", len(df))
```
